```python
FOTMOB_ID = "4653857"      # fotmob.com/match/<id>
WHOSCORED_ID = "2007643"   # whoscored.com/matches/<id>/live
OUT = "output"
CACHE = "output/raw"       # raw downloads are kept here (re-runs without internet)

from pmd.config import MatchConfig
from pmd.run import run
paths = run(MatchConfig(fotmob_match_id=FOTMOB_ID, whoscored_match_id=WHOSCORED_ID, out_dir=OUT, cache_dir=CACHE))
```

```python
from IPython.display import Image, display
import pandas as pd
display(Image(filename=[p for p in paths if p.endswith('.png')][0], width=900))
pd.read_csv([p for p in paths if p.endswith('_match_info.csv')][0])
```
